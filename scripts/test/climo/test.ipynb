```python
from dask_jobqueue import SLURMCluster
from dask.distributed import Client

cluster = SLURMCluster(
    queue="icelake",
    cores=2,
    memory="4GB",
    processes=1,
    walltime="00:10:00",
    interface="eno8303",  # Ensure correct networking
    job_extra_directives=[
        "--cpus-per-task=2",
        "--mem=4G",
        "--export=ALL",
    ],
    log_directory="/home/crodell/dask_logs",  # Store logs for debugging
)

# 🔹 Force submit jobs manually
cluster.scale(jobs=2)

# 🔹 Check if jobs are submitted
import time
time.sleep(5)  # Wait for SLURM to start jobs
print("Checking SLURM jobs...")

```

```python
!squeue -u $USER
```

```python

# 🔹 Start Dask client
client = Client(cluster)
print(client)

# 🔹 Wait for workers with timeout
try:
    client.wait_for_workers(2, timeout=300)  # 5-minute timeout
    print("Workers successfully started!")
except TimeoutError:
    print("Workers failed to connect after 5 minutes.")
```
